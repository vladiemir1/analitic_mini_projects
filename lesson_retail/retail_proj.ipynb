```python
import pandas as pd
import seaborn as sns
import matplotlib as plt
```

```python
retail = pd.read_csv(r'C:\Users\Владимир\Desktop\lesson_5\data.csv.zip',compression = 'zip', encoding = 'ISO-8859-1')
```

```python
retail_columns = retail.columns
retail.drop_duplicates(inplace = True)
```

```python
retail['InvoiceNo'].str.startswith('C').sum()
```

```python
retail = retail.query('Quantity > 0')
```

```python
germany_top = retail.query('Country == "Germany"').groupby('CustomerID',as_index = False)\
      .agg({'InvoiceNo':pd.Series.nunique}).sort_values('InvoiceNo',ascending = False)\
      .query('InvoiceNo > 7').reset_index().CustomerID
      
```

```python
top_retail_germany = retail[retail.CustomerID.isin(germany_top)]
```

```python
top_retail_germany.StockCode.value_counts().head(2)
```

```python
retail['Revenue'] = retail['CustomerID'] * retail['Quantity']
```

```python
retail.groupby('InvoiceNo',as_index = False).agg({'Revenue':'sum'}).sort_values('Revenue', ascending = False ).InvoiceNo.head(5)
```
